### Run a prompt against a local Ollama model

This cell sends a prompt to a locally running Ollama model and prints the response.

Before running:

1. Make sure Ollama is running:
   ollama serve

2. Make sure the model exists:
   ollama pull ministral-3

You can change:
- `model` → your preferred local model
- `content` → your prompt


```python
# Import the chat function (used to send prompts to Ollama models)
# and ChatResponse for optional type hinting
from ollama import chat
from ollama import ChatResponse

# Send a chat request to the local Ollama server
# model='ministral-3' -> change this to any model you have pulled (run: ollama list)
# messages -> conversation history (role can be 'user', 'system', or 'assistant')
response: ChatResponse = chat(model='ministral-3', messages=[
  {
    'role': 'user',
    'content': 'What is the color of sky?',
  },
])

# Print only the assistant's reply text from the response dictionary
print(response['message']['content'])

# Alternative way to access the same content (object-style)
# Uncomment if you prefer this syntax:
#print(response.message.content)
```

```python
# Send a chat request to your local Ollama model
# Make sure Ollama is running first:
#   ollama serve
#
# You can list available models with:
#   ollama list

response: ChatResponse = chat(
    model='ministral-3',  # Change this to any model you have installed locally
    messages=[
        {
            'role': 'user',
            # Replace this text with any question or instruction you want
            'content': 'Suggest me a Hindi movie to watch',
        },
    ]
)

# Print only the assistant's reply text
# response.message.content is equivalent to response['message']['content']
print(response.message.content)
```

### Creative generation with temperature

This cell asks the local LLM to generate creative text (a song).

Key parameters:

- `model` → your Ollama model name
- `content` → your prompt
- `temperature` → controls creativity

Temperature guide:

- 0.1–0.3 : factual / stable
- 0.5–0.7 : balanced
- 0.8–1.0 : highly creative

Make sure Ollama is running:

ollama serve


```python
# Send a creative prompt to the local Ollama model
# This example demonstrates how to control creativity using "temperature"

response: ChatResponse = chat(
    model='ministral-3',  # Change to any locally installed model (check with: ollama list)

    messages=[
        {
            'role': 'user',

            # User prompt:
            # Replace this with any instruction or question
            'content': 'write a song describing the beauty of a forest in Karnataka',
        },
    ],

    # Generation options passed to the model
    options={
        # Temperature controls randomness / creativity:
        # Lower values (0.1–0.3) → more factual & deterministic
        # Medium values (0.5–0.7) → balanced
        # Higher values (0.8–1.0+) → more creative / unpredictable
        "temperature": 0.7
    }
)

# Print the generated response text
print(response.message.content)
```

### Translation using LangChain + Ollama

This cell demonstrates:

- Creating a LangChain LLM using Ollama
- Building a prompt template with variables
- Injecting user values into the template
- Sending the formatted prompt to the model

Editable parts:

- model → your Ollama model name
- temperature → creativity level
- sentence → text to translate
- language → target language

The following is the classic Langchain flow:
Prompt Template -> Variables injected -> Formatted messages -> ChatOllama (local LLM) -> response.content

```python
# Import LangChain prompt templating utility
from langchain_core.prompts import ChatPromptTemplate

# Import Ollama chat wrapper for LangChain
from langchain_ollama import ChatOllama


# Create a LangChain LLM object backed by local Ollama
# model -> your locally installed Ollama model (check with: ollama list)
# temperature -> controls creativity (lower = more deterministic)
llm = ChatOllama(
    model="ministral-3",
    temperature=0.7
)


# Define a reusable chat prompt template
# "system" message sets the assistant's role/behavior
# "user" message contains placeholders {sentence} and {language}
prompt_template = ChatPromptTemplate.from_messages(
    [
        ("system", "You are language translator"),
        ("user", "Translate this {sentence} to {language}")
    ]
)


# Fill the template variables with actual values
# This produces formatted chat messages for the LLM
messages = prompt_template.invoke({
    "sentence": "Je suis artiste",
    "language": "English"
})


# Send the formatted messages to the model
response = llm.invoke(messages)


# Print only the model's generated text
print(response.content)
```

### Translate another sentence (reuse existing prompt)

This cell demonstrates reusing the same prompt template and LLM
with different input text.

Only update:

- `sentence`
- `language`

Everything else remains unchanged.

```python
# Reuse the existing prompt_template to translate a new sentence
# Only the input variables change — no need to recreate the template or LLM

messages = prompt_template.invoke({
    # Source sentence (Telugu in this example)
    "sentence": "నాన్నకు ప్రేమతో",

    # Target language
    "language": "English"
})

# Send the formatted messages to the Ollama-backed LangChain LLM
response = llm.invoke(messages)

# Print the translated result
print(response.content)
```

### Simple prompt template with one variable

This example uses a single-variable prompt:

- `{topic}` is replaced at runtime
- The formatted prompt is sent to the LLM
- The response is printed

Change `topic` to query anything.

```python
# Create a simple prompt template with a variable {topic}
# This is a single-message template (not multi-role like before)
prompt_template = ChatPromptTemplate.from_template(
    "Tell me in short about {topic}"
)

# Inject a value into {topic}
# This produces a formatted prompt object ready for the LLM
prompt = prompt_template.invoke({
    "topic": "butter"
})

# Send the prompt to the LangChain Ollama model
response = llm.invoke(prompt)

# Print only the generated text
print(response.content)
```

### Prompt → LLM chaining using pipe operator

The `|` operator connects components into a pipeline:

PromptTemplate | LLM

Now you only call `.invoke()` once instead of two steps.

```python
# Create a chain by piping the prompt template directly into the LLM
# This combines:
# PromptTemplate → LLM
topic_chain = prompt_template | llm

# Invoke the chain with input variables
# This internally:
# 1. Fills the template
# 2. Sends to the model
# 3. Returns the AI response
topic_chain.invoke({
    "topic": "butter"
})
```

## Output Parsing (Return Plain Strings)

By default, LangChain returns `AIMessage` objects.

Here we add `StrOutputParser()` so `.invoke()` returns a simple Python string instead.
This makes outputs easier to print, store, or post-process.

```python
from langchain_core.output_parsers import StrOutputParser

# Build a chain:
# 1) Fill the prompt template
# 2) Call the LLM
# 3) Parse the output into a plain Python string (instead of AIMessage)
topic_chain = prompt_template | llm | StrOutputParser()

# Now invoke returns a string directly
topic_chain.invoke({"topic": "butter"})
```

## Movie Title Generator (Prompt → LLM → String)

This chain demonstrates:

- Using multiple input variables (`language`, `genre`)
- Prompt templating
- Parsing LLM output into plain text

You can change:
- `language`
- `genre`
- Prompt wording

```python
# Prompt template with two variables: {language} and {genre}
movie_title_template = ChatPromptTemplate.from_template(
    "Suggest me 1 movie title which is in the language : {language} and genre: {genre}."
)

# Chain that returns a plain string movie title
movie_title_chain = movie_title_template | llm | StrOutputParser()

# Provide values for both variables
movie_title_chain.invoke({"language": "hindi", "genre": "horror"})
```

## Chain Composition (Title → Summary)

This example shows how to:

1. Generate a movie title
2. Pass that title into another prompt
3. Generate a short summary

We map the output of one chain into `{title}` for the next chain.

This is a common LangChain pattern for multi-step workflows.

```python
from langchain_core.runnables import RunnableLambda

# Small passthrough runnable (here it doesn't transform anything)
# You can later replace this to log, clean text, or validate output.
print_movie = RunnableLambda(lambda x: x)

# Prompt that expects a {title} input
movie_summary_prompt = ChatPromptTemplate.from_template(
    "Give me 2–3 sentence summary of the movie with title: {title}"
)

# Compose a multi-step chain:
# Step A: generate a title (movie_title_chain)
# Step B: feed that result into the next prompt as {"title": ...}
# Step C: ask LLM for a summary
# Step D: parse output into plain string
composed_movie_chain = (
    {"title": movie_title_chain}      # map output into the key "title"
    | print_movie                     # optional passthrough step
    | movie_summary_prompt            # builds a prompt using {title}
    | llm
    | StrOutputParser()
)

# Only need to give inputs required by movie_title_chain
composed_movie_chain.invoke({"language": "english", "genre": "horror"})
```

## RunnableSequence (Explicit Pipeline)

Same logic as before, but written using `RunnableSequence`
to make each step in the pipeline more explicit and readable.

Use this style when chains become long or complex.

```python
from langchain_core.runnables import RunnableSequence

# RunnableSequence is the explicit “pipeline” form of chaining.
# It can be easier to read when chains get long.
composed_movie_chain = RunnableSequence(
    {"title": movie_title_chain},  # generate title first
    print_movie,                   # optional passthrough/logging
    movie_summary_prompt,           # create summary prompt using {title}
    llm,
    StrOutputParser()              # return plain string
)

composed_movie_chain.invoke({"language": "english", "genre": "horror"})
```

## RunnableParallel (Multiple Outputs at Once)

This section demonstrates running multiple chains simultaneously.

Here:

- One chain translates to Hindi
- Another translates to French

Both receive the same input sentence and return results as a dictionary.

This is useful for:
- Multi-language generation
- Multiple formats
- Parallel evaluations

```python
from langchain_core.runnables import RunnableParallel

# Build a Hindi translation chain
translate_hindi_prompt = ChatPromptTemplate.from_template(
    "Translate this sentence to Hindi : {sentence}"
)
translate_hindi_chain = translate_hindi_prompt | llm | StrOutputParser()

# Build a French translation chain
translate_french_prompt = ChatPromptTemplate.from_template(
    "Translate this sentence to French : {sentence}"
)
translate_french_chain = translate_french_prompt | llm | StrOutputParser()

# RunnableParallel runs multiple chains at the same time on the same input
# and returns a dictionary of results.
translate_runnable = RunnableParallel(
    hindi_translate=translate_hindi_chain,
    french_translate=translate_french_chain
)

# One input → two outputs
translations = translate_runnable.invoke({"sentence": "i want to swim in the river"})

# Access results by the keys used in RunnableParallel(...)
print(f"hindi summary: {translations['hindi_translate']}")
print(f"french summary: {translations['french_translate']}")
```
